Muestra nombre y precio de todos los productos, del mas caro al mas barato

```python
--EJERCISIO 1
SELECT nombre, precio
FROM productos
ORDER BY precio DESC;
```

```python
--EJERCISIO 2
SELECT *
FROM clientes
WHERE pais IN ('ES', 'MX')
  AND nombre LIKE 'M%';
```

```python
--EJERCISIO 3
SELECT *
FROM productos
WHERE CAST(precio AS DOUBLE) BETWEEN 20 AND 60
  AND stock > 0;
```

```python
--EJERCISIO 4
SELECT nombre, COALESCE(telefono, 'sin dato') AS telefono
FROM clientes;
```

```python
--EJERCISIO 5
SELECT
    nombre,
    precio,
    CAST(precio AS DOUBLE) * 1.21 AS precio_con_iva
FROM productos
ORDER BY CAST(precio AS DOUBLE) DESC
LIMIT 10;
```

```python
--EJERCISIO 6
SELECT
    nombre,
    precio,
    CASE
        WHEN CAST(precio AS DOUBLE) < 20 THEN 'barato'
        WHEN CAST(precio AS DOUBLE) <= 100 THEN 'medio'
        ELSE 'caro'
    END AS categoria_precio
FROM productos;
```

```python
--EJERCISIO 7
SELECT COUNT(DISTINCT pais) AS paises_distintos
FROM clientes;
```
