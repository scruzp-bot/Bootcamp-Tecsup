```python
#Ejercicio 1
#Dada la tupla numeros = (15, 8, 42, 3, 27, 19, 5), obtén el número mayor, el menor y la diferencia entre ambos. 
tupla_numeros = (15, 8, 42, 3, 27, 19, 5)
numero_mayor = max(tupla_numeros)
numero_menor = min(tupla_numeros)
diferencia = numero_mayor - numero_menor
print("Número mayor:", numero_mayor)
print("Número menor:", numero_menor)
print("Diferencia:", diferencia)
```

```python
#Ejercicio 2
#Dada la tupla datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3), solicita un número al usuario y muestra cuántas veces aparece sin convertir la tupla en lista.
datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3)
numero = int(input("Ingrese un número: "))
veces = datos.count(numero)
print(f"El número {numero} aparece {veces} veces en la tupla.")
```

```python
#Ejercicio 3
#Dada la tupla nombres = ('Ana','Luis','Pedro','María','Luis','Carlos'), solicita un nombre y 
#muestra su primera posición o un mensaje si no existe. 
nombres = ('Ana','Luis','Pedro','María','Luis','Carlos')
nombre = input("Ingrese un nombre: ")
if nombre in nombres:
    posicion = nombres.index(nombre)
    print(f"El nombre '{nombre}' se encuentra en la posición {posicion}.")
```

```python
#Ejercicio 4
#Con notas = (18,15,20,11,14,17,19), calcula promedio, máxima, mínima y cantidad de aprobados
notas = (18,15,20,11,14,17,19)
promedio = sum(notas) / len(notas)
print("Promedio:", promedio)
nota_maxima = max(notas)
print("Nota máxima:", nota_maxima)
nota_minima = min(notas)
print("Nota mínima:", nota_minima)
aprobados = len([nota for nota in notas if nota >= 15])
print("Cantidad de aprobados:", aprobados)
```

```python
#Ejercicio 5
#Con ventas=[('Lunes',120),('Martes',150),('Miércoles',95),('Jueves',180),('Viernes',210)], calcula el día con mayor venta, el total y el promedio. 
ventas = [('Lunes',120),('Martes',150),('Miércoles',95),('Jueves',180),('Viernes',210)]
total_ventas = sum(venta[1] for venta in ventas)
promedio_ventas = total_ventas / len(ventas)
dia_mayor_venta = max(ventas, key=lambda x: x[1])
print("Día con mayor venta:", dia_mayor_venta[0], "con", dia_mayor_venta[1], "ventas.")
print("Total de ventas:", total_ventas)
print("Promedio de ventas:", promedio_ventas)
```
