```python
# Ejercicio 1
ventas = [150.50, 200, 80.90, 300, 175.50, 90]


def analizar_ventas(ventas):
    if not ventas or any(v < 0 for v in ventas):
        return "Error: La lista de ventas está vacía o contiene valores negativos."
    
    total_vendido = sum(ventas)
    promedio = total_vendido / len(ventas)
    venta_mas_alta = max(ventas)
    venta_mas_baja = min(ventas)
    ventas_superiores_promedio = sum(1 for v in ventas if v > promedio)
    ventas_por_debajo_de_100 = sum(1 for v in ventas if v < 100)

    resultados = {
        "total_vendido": total_vendido,
        "promedio": promedio,
        "venta_mas_alta": venta_mas_alta,
        "venta_mas_baja": venta_mas_baja,
        "ventas_superiores_promedio": ventas_superiores_promedio,
        "ventas_por_debajo_de_100": ventas_por_debajo_de_100
    }
    
    return resultados
analizar_ventas(ventas)

```

```python
#Ejercicio 2


def validar_password(password):
    errores = []
    
    if len(password) < 8:
        errores.append("La contraseña debe tener al menos 8 caracteres.")
    if not any(c.isupper() for c in password):
        errores.append("La contraseña debe contener al menos una letra mayúscula.")
    if not any(c.islower() for c in password):
        errores.append("La contraseña debe contener al menos una letra minúscula.")
    if not any(c.isdigit() for c in password):
        errores.append("La contraseña debe contener al menos un número.")
    if not any(c in '@#$%' for c in password):
        errores.append("La contraseña debe incluir al menos uno de estos símbolos: @, #, $ o %.")
    if ' ' in password:
        errores.append("La contraseña no debe contener espacios.")
    
    es_valida = len(errores) == 0
    return es_valida, errores

es_valida, errores = validar_password('Python123')

print(es_valida) 
print(errores) 
```

```python
#Ejercicio 3




def registrar_estudiante(estudiantes, nombre, notas):
    if not nombre:
        return "Error: El nombre del estudiante no puede estar vacío."
    if nombre in estudiantes:
        return "Error: El estudiante ya está registrado."
    if len(notas) < 3:
        return "Error: El estudiante debe tener al menos tres notas."
    if any(not isinstance(nota, (int, float)) or nota < 0 or nota > 20 for nota in notas):
        return "Error: Todas las notas deben ser numéricas y estar entre 0 y 20."
    
    promedio = round(sum(notas) / len(notas), 2)
    estado = "Aprobado" if promedio >= 11 else "Desaprobado"
    
    estudiantes[nombre] = {
        "notas": notas,
        "promedio": promedio,
        "estado": estado
    }
    
    return f"Registro exitoso del estudiante {nombre}."
estudiantes = {}
mensaje = registrar_estudiante(estudiantes, "Juan", [15, 18, 12])
print(mensaje)
print(estudiantes)

mensaje = registrar_estudiante(estudiantes, "Ana", [10, 9, 8])
print(mensaje)
print(estudiantes)
```

```python
#Ejercicio 4


def agrupar_por_longitud(palabras):
    resultado = {}
    palabras_unicas = set(palabra.lower() for palabra in palabras if isinstance(palabra, str))
    
    for palabra in palabras_unicas:
        longitud = len(palabra)
        if longitud not in resultado:
            resultado[longitud] = []
        resultado[longitud].append(palabra)
    
    for longitud in resultado:
        resultado[longitud].sort()
    
    return resultado

palabras = [ 
'python', 'Java', 'código', 'web', 
'java', 'datos', 'API', 'lista' 
] 

resultado = agrupar_por_longitud(palabras)
print(resultado)
```

```python
#Ejercicio 5


def actualizar_inventario(inventario, producto, cantidad, operacion):
    if operacion not in ["entrada", "salida"]:
        return {"mensaje": "Error: La operación debe ser 'entrada' o 'salida'.", "stock_actual": inventario.get(producto, 0)}
    if not isinstance(cantidad, int) or cantidad <= 0:
        return {"mensaje": "Error: La cantidad debe ser un entero positivo.", "stock_actual": inventario.get(producto, 0)}
    
    if operacion == "entrada":
        inventario[producto] = inventario.get(producto, 0) + cantidad
        return {"mensaje": f"Entrada registrada para {producto}.", "stock_actual": inventario[producto]}
    
    elif operacion == "salida":
        if producto not in inventario:
            return {"mensaje": f"Error: El producto {producto} no existe en el inventario.", "stock_actual": 0}
        if inventario[producto] < cantidad:
            return {"mensaje": f"Error: Stock insuficiente para {producto}.", "stock_actual": inventario[producto]}
        
        inventario[producto] -= cantidad
        return {"mensaje": f"Salida registrada para {producto}.", "stock_actual": inventario[producto]}
inventario = { 
'Laptop': 5, 
'Mouse': 12, 
'Teclado': 8 
}

actualizar_inventario( 
inventario, 'Mouse', 4, 'salida' 
) 

actualizar_inventario(inventario, 'Monitor', 3, 'entrada')
inventario

actualizar_inventario(inventario, 'Laptop', 6, 'salida')
inventario
```

```python
#Ejercicio 6


def calcular_pedido(productos, cantidades, descuento=0, delivery=0):
    if not (0 <= descuento <= 100):
        return {"mensaje": "Error: El descuento debe estar entre 0 y 100."}
    
    subtotal = 0
    no_encontrados = []
    
    for producto, cantidad in cantidades.items():
        if producto not in productos:
            no_encontrados.append(producto)
            continue
        if not isinstance(cantidad, int) or cantidad <= 0:
            return {"mensaje": f"Error: La cantidad de {producto} debe ser un entero positivo."}
        
        subtotal += productos[producto] * cantidad
    
    descuento_monto = subtotal * (descuento / 100)
    total_final = subtotal - descuento_monto + delivery
    
    resultado = {
        "subtotal": round(subtotal, 2),
        "descuento": round(descuento_monto, 2),
        "delivery": round(delivery, 2),
        "total_final": round(total_final, 2),
        "no_encontrados": no_encontrados
    }
    
    return resultado

productos = {'Hamburguesa': 15, 'Pizza': 25, 'Gaseosa': 5} 
cantidades = {'Hamburguesa': 2, 'Gaseosa': 3, 'Helado': 1} 

resultado = calcular_pedido(productos, cantidades, descuento=10, delivery=5)
print(resultado)
```

```python
#Ejercicio 7

def evaluar_empleados(empleados, minimo=70):
    resultados = []
    
    for nombre, puntajes in empleados.items():
        promedio = sum(puntajes) / len(puntajes)
        
        if promedio >= minimo:
            if promedio >= 90:
                clasificacion = "Excelente"
            elif promedio >= 80:
                clasificacion = "Bueno"
            elif promedio >= 70:
                clasificacion = "Regular"
            else:
                clasificacion = "Deficiente"
            
            resultados.append((nombre, round(promedio, 2), clasificacion))
    
    resultados.sort(key=lambda x: (-x[1], x[0]))
    
    return resultados

empleados = { 
'Ana': [80, 90, 75], 
'Carlos': [60, 70, 65], 
'Lucía': [95, 85, 90], 
'Pedro': [70, 70, 70] 
} 

resultado = evaluar_empleados(empleados, minimo=70)
print(resultado)
```

```python
#Ejercicio 8



def comprimir_texto(texto):
    if not texto:
        return ""
    
    comprimido = []
    contador = 1
    
    for i in range(1, len(texto)):
        if texto[i] == texto[i - 1]:
            contador += 1
        else:
            comprimido.append(f"{texto[i - 1]}{contador}")
            contador = 1
    
    comprimido.append(f"{texto[-1]}{contador}")
    
    return ''.join(comprimido)

def descomprimir_texto(texto):
    if not texto:
        return ""
    
    descomprimido = []
    i = 0
    
    while i < len(texto):
        if not texto[i].isalpha():
            return "Error: Formato comprimido inválido."
        
        caracter = texto[i]
        i += 1
        cantidad_str = ""
        
        while i < len(texto) and texto[i].isdigit():
            cantidad_str += texto[i]
            i += 1
        
        if not cantidad_str:
            return "Error: Formato comprimido inválido."
        
        try:
            cantidad = int(cantidad_str)
        except ValueError:
            return "Error: Formato comprimido inválido."
        
        descomprimido.append(caracter * cantidad)
    
    return ''.join(descomprimido)

texto = 'aaabbccccdaa' 
comprimido = comprimir_texto(texto) 
original = descomprimir_texto(comprimido) 

print(f"Texto original: {texto}")
print(f"Texto comprimido: {comprimido}")
print(f"Texto descomprimido: {original}")
print(descomprimir_texto('a3b2c4d1a2'))  # Ejemplo de descompresión
```

```python
#Ejercicio 9


def transferir(cuentas, origen, destino, monto):
    if origen not in cuentas:
        return {"estado": "Error", "mensaje": f"La cuenta de origen {origen} no existe."}
    if destino not in cuentas:
        return {"estado": "Error", "mensaje": f"La cuenta de destino {destino} no existe."}
    if origen == destino:
        return {"estado": "Error", "mensaje": "Las cuentas de origen y destino deben ser diferentes."}
    if not isinstance(monto, (int, float)) or monto <= 0:
        return {"estado": "Error", "mensaje": "El monto debe ser numérico y mayor que cero."}
    if cuentas[origen] < monto:
        return {"estado": "Error", "mensaje": "Saldo insuficiente en la cuenta de origen."}
    
    cuentas[origen] -= monto
    cuentas[destino] += monto
    
    comprobante = {
        "estado": "Éxito",
        "origen": origen,
        "destino": destino,
        "monto": monto,
        "saldo_origen_restante": cuentas[origen],
        "saldo_destino_actualizado": cuentas[destino]
    }
    
    return comprobante

cuentas = { 
'001': {'titular': 'Ana', 'saldo': 1500}, 
'002': {'titular': 'Luis', 'saldo': 800} 
} 

transferir ('cuentas', '001', '002', 300) 
print(cuentas)
```
