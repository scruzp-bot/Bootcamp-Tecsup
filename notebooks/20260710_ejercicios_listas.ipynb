```python
#Ejercicio 1
#Dada una lista con números repetidos, crea una nueva lista sin números repetidos. 
numeros = [4, 7, 2, 4, 9, 7, 1, 2, 10, 4] 
numeros_sin_repetir = list(set(numeros))
print(numeros_sin_repetir)
```

```python
#EJERCICIO 2
#Dada una lista de números, crea dos listas: una con números pares y otra con números impares. 
numeros = [12, 7, 9, 24, 31, 18, 5, 42, 13, 8] 
numeros_pares = [num for num in numeros if num % 2 == 0]
numeros_impares = [num for num in numeros if num % 2 != 0]
print("Números pares:", numeros_pares)
print("Números impares:", numeros_impares)
```

```python
#Ejercisio 3
#Dada una lista de números, encuentra el segundo valor más grande. 
numeros = [45, 12, 89, 34, 89, 23, 67, 10]
numeros_distinct = []
for numero in numeros:
    if numero not in numeros_distinct:
        numeros_distinct.append(numero)

numeros_distinct.sort(reverse=True)
print(f"El segundo mas grande es: {numeros_distinct[1]}")
```

```python
#Ejercisio 4
#Dada una lista de palabras, cuenta cuántas palabras tienen más de 5 letras. 
palabras = ['python', 'sol', 'programacion', 'lista', 'computadora', 'if', 'variable'] 
palabras_largas = [palabra for palabra in palabras if len(palabra) > 5]
print(f"Número de palabras con más de 5 letras: {len(palabras_largas)}")
print("Palabras con más de 5 letras:", palabras_largas)
```

```python
#Ejercisio 5
#Dada una lista de notas, calcula el promedio e indica si el estudiante aprobó o desaprobó. 
notas = [14, 18, 11, 16, 9, 13, 20] 
promedio = sum(nota for nota in notas) / len(notas)
if promedio >= 13:
    print(f"Promedio: {promedio:.2f}. El estudiante aprobó.")
else:
    print(f"Promedio: {promedio:.2f}. El estudiante desaprobó.")
```

```python
# Ejercicio 6
#Dada una lista con números positivos y negativos, crea una nueva lista solo con los números positivos. 
numeros = [-5, 8, -2, 15, 0, -9, 23, -1, 7]
numeros_positivos = [num for num in numeros if num > 0]
print("Números positivos:", numeros_positivos)
```

```python
# Ejercicio 7
#Dada una lista de números, crea otra lista donde cada número esté multiplicado por 2. 
numeros = [3, 6, 9, 12, 15]
numeros_multiplicados = [num * 2 for num in numeros]
print("Números multiplicados por 2:", numeros_multiplicados)
```

```python
#Ejercisio 8
#Dadas dos listas, encuentra los elementos que aparecen en ambas. 
lista1 = [1, 3, 5, 7, 9, 11] 
lista2 = [2, 3, 6, 7, 10, 11] 
elementos_comunes = [num for num in lista1 if num in lista2]
print("Elementos comunes:", elementos_comunes)
```

```python
#Ejercisio 9
#Dadas dos listas, crea una nueva lista con los elementos que no se repiten entre ambas listas. 
lista1 = [4, 8, 12, 16, 20] 
lista2 = [8, 10, 12, 18, 20] 
elementos_no_repetidos = [num for num in lista1 + lista2 if (num in lista1) ^ (num in lista2)]
print("Elementos no repetidos:", elementos_no_repetidos)
```

```python
#Ejercisio 10
#Dada una lista de colores, invierte la lista sin usar el método .reverse(). 
colores = ['rojo', 'azul', 'verde', 'amarillo', 'morado'] 
colores_invertidos = colores[::-1]
print("Colores invertidos:", colores_invertidos)
```

```python
#Ejercisio 11
#Dada una lista de palabras, ordénalas desde la más corta hasta la más larga. 
palabras = ['casa', 'elefante', 'sol', 'computadora', 'mesa', 'programacion'] 
longitud_palabras = [(palabra, len(palabra)) for palabra in palabras]
longitud_palabras.sort(key=lambda x: x[1])
palabras_ordenadas = [palabra for palabra, _ in longitud_palabras]
print("Palabras ordenadas por longitud:", palabras_ordenadas)
```

```python
#Ejercisio 12
#Dada una lista con textos y strings vacíos, crea una nueva lista eliminando los elementos vacíos. 
textos = ['hola', '', 'python', '', 'listas', ' ', 'programar', '']
textos_no_vacios = [texto for texto in textos if texto.strip() != '']
print("Textos sin elementos vacíos:", textos_no_vacios)
```

```python
#Ejercisio 13
#Dada una lista de nombres repetidos, muestra cuántas veces aparece cada nombre. 
nombres = ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'María', 'Carlos']
nombre_unicos = set(nombres)
conteo_nombres = {nombre: nombres.count(nombre) for nombre in nombre_unicos}
print("Conteo de nombres:", conteo_nombres)
```

```python
#Ejercisio 15
#Dada una lista de frutas, pide al usuario una fruta y muestra su posición si existe en la lista. 
frutas = ['manzana', 'pera', 'uva', 'sandía', 'plátano', 'mango'] 
fruta = input("Ingrese una fruta: ")
if fruta in frutas:
    posicion = frutas.index(fruta)
    print(f"La fruta '{fruta}' se encuentra en la posición {posicion}.")
else:
    print(f"La fruta '{fruta}' no se encuentra en la lista.")
```

```python
#Ejercisio 16
#Dada una lista de números, divídela en dos mitades. 
numeros = [10, 20, 30, 40, 50, 60, 70, 80] 
mitad1 = numeros[:len(numeros)//2]
mitad2 = numeros[len(numeros)//2:]
print("Mitad 1:", mitad1)
print("Mitad 2:", mitad2)
```

```python
#Ejercisio 17
#Dada una lista de palabras, reemplaza todas las apariciones de 'java' por 'python'. 
palabras = ['java', 'python', 'java', 'c++', 'java', 'javascript'] 
palabras_reemplazadas = ['python' if palabra == 'java' else palabra for palabra in palabras]
print("Palabras con 'java' reemplazado por 'python':", palabras_reemplazadas)
```

```python
#Ejercisio 19
#Dadas dos listas del mismo tamaño, crea una nueva lista alternando los elementos de ambas listas. 
numeros = [1, 2, 3, 4] 
letras = ['a', 'b', 'c', 'd'] 
resultado_esperado = [1, 'a', 2, 'b', 3, 'c', 4, 'd'] 
resultado = [None] * (len(numeros) + len(letras))
for i in range(len(numeros)):
    resultado[2 * i] = numeros[i]
    resultado[2 * i + 1] = letras[i]
print("Resultado alternando elementos:", resultado)
```

```python
#Ejercisio 20
#Dada una lista de estudiantes y otra lista con sus notas, crea una nueva lista solo con los nombres de los estudiantes que aprobaron. Considera que la nota mínima aprobatoria es 11. 
estudiantes = ['Ana', 'Luis', 'Carlos', 'María', 'Pedro'] 
notas = [18, 10, 14, 8, 16] 
estudiantes_aprobados = [estudiantes[i] for i in range(len(estudiantes)) if notas[i] >= 11]
print("Estudiantes aprobados:", estudiantes_aprobados)
```
