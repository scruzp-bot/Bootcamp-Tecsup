```python
#Ejercicio 1

estudiantes_python = { 
"Ana", "Luis", "María", "Carlos", "Sofía", "Diego" 
} 

estudiantes_javascript = { 
"María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria" 
} 

# Los estudiantes inscritos en ambos cursos
estudiantes_ambos = estudiantes_python.intersection(estudiantes_javascript)
print("Estudiantes inscritos en ambos cursos:", estudiantes_ambos)

# Los estudiantes que solo están en Python
estudiantes_solo_python = estudiantes_python.difference(estudiantes_javascript)
print("Estudiantes solo en Python:", estudiantes_solo_python)

# Los estudiantes que solo están en JavaScript
estudiantes_solo_javascript = estudiantes_javascript.difference(estudiantes_python)
print("Estudiantes solo en JavaScript:", estudiantes_solo_javascript)

# Todos los estudiantes sin repetir nombres
todos_estudiantes = estudiantes_python.union(estudiantes_javascript)
print("Todos los estudiantes:", todos_estudiantes)
```

```python
# Ejercicio 2
almacen_arequipa = {
    "laptop", "teclado", "mouse", "monitor", "impresora", "audifonos"
}

almacen_lima = {
    "laptop", "mouse", "monitor", "tablet", "webcam", "parlantes"
}

# Determina
# Los productos disponibles en ambos almacenes. 
productos_comunes = almacen_arequipa.intersection(almacen_lima)
print("Productos disponibles en ambos almacenes:", productos_comunes)

# Los productos exclusivos del almacén de Arequipa. 
productos_arequipa = almacen_arequipa.difference(almacen_lima)
print("Productos exclusivos del almacén de Arequipa:", productos_arequipa)

#Los productos exclusivos del almacén de Lima. 
productos_lima = almacen_lima.difference(almacen_arequipa)
print("Productos exclusivos del almacén de Lima:", productos_lima)

# Todos los productos disponibles sin repetir. 
todos_productos = almacen_arequipa.union(almacen_lima)
print("Todos los productos disponibles sin repetir:", todos_productos)

# Si ambos almacenes tienen exactamente el mismo inventario.
inventarios_iguales = almacen_arequipa == almacen_lima
print("¿Ambos almacenes tienen exactamente el mismo inventario?", inventarios_iguales)
```

```python
# Ejercicio 3
permisos_usuario = {
    "leer", "editar", "comentar", "descargar"
}

permisos_requeridos = {
    "leer", "editar", "eliminar", "publicar"
}

#Crea un programa que muestre: 

# Si el usuario posee todos los permisos requeridos. 

permisos_completos = permisos_requeridos.issubset(permisos_usuario)
print("¿El usuario posee todos los permisos requeridos?", permisos_completos)

#Los permisos que le faltan. 

permisos_faltantes = permisos_requeridos.difference(permisos_usuario)
print("Permisos que le faltan al usuario:", permisos_faltantes)

# Los permisos adicionales que tiene. 
permisos_adicionales = permisos_usuario.difference(permisos_requeridos)
print("Permisos adicionales que tiene el usuario:", permisos_adicionales)

# La cantidad total de permisos diferentes.

cantidad_permisos_diferentes = len(permisos_usuario.union(permisos_requeridos))
print("Cantidad total de permisos diferentes:", cantidad_permisos_diferentes)
```

```python
# Ejercicio 4
numeros_a = {2, 4, 6, 8, 10, 12, 14}
numeros_b = {4, 6, 9, 10, 13, 14, 16}
numeros_c = {1, 4, 6, 10, 11, 14, 18}


#Encuentra

# Los números presentes en los tres conjuntos.

numeros_comunes = numeros_a.intersection(numeros_b).intersection(numeros_c)
print("Números presentes en los tres conjuntos:", numeros_comunes)

#Los números presentes en numeros_a y numeros_b, pero no en numeros_c. 

numeros_a_b_no_c = numeros_a.intersection(numeros_b).difference(numeros_c)
print("Números presentes en numeros_a y numeros_b, pero no en numeros_c:", numeros_a_b_no_c)

#Los números que solo pertenecen a numeros_a. 

numeros_solo_a = numeros_a.difference(numeros_b).difference(numeros_c)
print("Números que solo pertenecen a numeros_a:", numeros_solo_a)

# Todos los números diferentes.

todos_numeros_diferentes = numeros_a.union(numeros_b).union(numeros_c)
print("Todos los números diferentes:", todos_numeros_diferentes)
```

```python
# Ejercicio 5
habilidades_candidato_1 = {
    "Python", "Git", "SQL", "Docker", "React"
}

habilidades_candidato_2 = {
    "Python", "JavaScript", "React", "Node.js", "Git"
}

habilidades_requeridas = {
    "Python", "Git", "SQL", "Docker"
}

#Determina
#Las habilidades que comparten ambos candidatos. 
habilidades_comunes = habilidades_candidato_1.intersection(habilidades_candidato_2)
print("Habilidades que comparten ambos candidatos:", habilidades_comunes)

#Las habilidades exclusivas de cada candidato. 

habilidades_exclusivas_candidato_1 = habilidades_candidato_1.difference(habilidades_candidato_2)
print("Habilidades exclusivas del candidato 1:", habilidades_exclusivas_candidato_1)


#Qué habilidades requeridas le faltan a cada candidato.

habilidades_faltantes_candidato_1 = habilidades_requeridas.difference(habilidades_candidato_1)
print("Habilidades requeridas que le faltan al candidato 1:", habilidades_faltantes_candidato_1)

#Qué candidato cumple con más requisitos. 

cumple_candidato_1 = len(habilidades_requeridas.intersection(habilidades_candidato_1))
cumple_candidato_2 = len(habilidades_requeridas.intersection(habilidades_candidato_2))

if cumple_candidato_1 > cumple_candidato_2:
    print("El candidato 1 cumple con más requisitos.")
elif cumple_candidato_2 > cumple_candidato_1:
    print("El candidato 2 cumple con más requisitos.")
else:
    print("Ambos candidatos cumplen con el mismo número de requisitos.")

# Todas las habilidades disponibles sin repetir. 

todas_habilidades = habilidades_candidato_1.union(habilidades_candidato_2).union(habilidades_requeridas)
print("Todas las habilidades disponibles sin repetir:", todas_habilidades)
```
