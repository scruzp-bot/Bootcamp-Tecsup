```python
# Crea un array con los números del 10 al 50 de 5 en 5.
array = np.arange(10, 50, 5)
print(array)
print("=" * 100)

# Imprime shape, ndim, dtype y size.
print(array.shape)
print(array.ndim)
print(array.dtype)
print(array.size)
print("=" * 100)

# Crea una matriz 3×4 llena de ceros y otra 3×3 identidad.
matriz_zeros = np.zeros((3, 4))
identidad = np.eye(3)
print(matriz_zeros)
print(identidad)
print("=" * 100)

# Genera 7 valores equiespaciados entre -1 y 1.
arr = np.linspace(-1, 1, 7)
print(arr)
print("=" * 100)

# Convierte el primer array a float32 y comprueba itemsize antes y después.
array_flotante = array.astype('float32')
print(array_flotante.dtype)
print(array_flotante.itemsize)
```
