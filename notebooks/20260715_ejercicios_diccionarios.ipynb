```python
# Ejercicio 5
#Agrupa empleados primero por departamento y luego por nivel de experiencia. 

empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700}
]

#Requisitos

# En cada grupo almacena nombres, cantidad y salario promedio. 

agrupados = {}
for empleado in empleados:
    depto = empleado["departamento"]
    nivel = empleado["nivel"]
    
    if depto not in agrupados:
        agrupados[depto] = {}
    
    if nivel not in agrupados[depto]:
        agrupados[depto][nivel] = {
            "nombres": [],
            "cantidad": 0,
            "salario_promedio": 0
        }
    
    agrupados[depto][nivel]["nombres"].append(empleado["nombre"])
    agrupados[depto][nivel]["cantidad"] += 1
    agrupados[depto][nivel]["salario_promedio"] += empleado["salario"]

print("Agrupación de empleados:", agrupados)

#La estructura debe ser un diccionario anidado. 

diccionario_anidado = agrupados
print("Diccionario anidado:", diccionario_anidado)
```

```python
# Ejercicio 7
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]

mapa = {
    "ana": "Ana",
    "luis": "Luis",
    "carla": "Carla"
}

lista_candidatos = list(candidatos)
lista_candidatos.sort()

# Los votos deben ser insensibles a mayúsculas. 

candidatos_minuscula = []
for candidato in lista_candidatos:
    candidatos_minuscula.append(candidato.lower())
conteo = {}
for candidato in lista_candidatos:
    conteo[candidato] = 0

votos_invalidos = []
for voto in votos:
    voto_minuscula = voto.lower()

    if voto_minuscula in candidatos_minuscula:
        # Buscamos a que candidato oficial corresponde
        for candidato in lista_candidatos:
            if candidato.lower() == voto_minuscula:
                conteo[candidato] = conteo[candidato] + 1
    else:
        votos_invalidos.append(voto)
total_validos = 0
for candidato in lista_candidatos:
    total_validos = total_validos + conteo[candidato]

print("CONTEO POR CANDIDATO")
print("-------------------------------")

for candidato in lista_candidatos:
    cantidad = conteo[candidato]

    if total_validos > 0:
        porcentaje = cantidad * 100 / total_validos
    else:
        porcentaje = 0

    print(candidato, ":", cantidad, "votos -", round(porcentaje, 2), "%")

print("-------------------------------")
print("Votos emitidos :", len(votos))
print("Votos validos  :", total_validos)
print("Votos invalidos:", len(votos_invalidos), "->", votos_invalidos)
print("-------------------------------")


if total_validos == 0:
    print("No hubo votos validos, no se puede definir ganador")
else:
    maximo = 0
    for candidato in lista_candidatos:
        if conteo[candidato] > maximo:
            maximo = conteo[candidato]

    ganadores = []
    for candidato in lista_candidatos:
        if conteo[candidato] == maximo:
            ganadores.append(candidato)

    if len(ganadores) == 1:
        porcentaje_ganador = maximo * 100 / total_validos
        print("GANADOR:", ganadores[0])
        print("Obtuvo", maximo, "votos (", round(porcentaje_ganador, 2), "% )")
    else:
        print("HAY EMPATE entre:", ganadores)
        print("Cada uno obtuvo", maximo, "votos")

```

```python
#Ejercisio 9
#Aplica una secuencia de movimientos de entrada y salida sobre un inventario. 
stock = {"laptop": 5, "mouse": 12, "monitor": 4} 
movimientos = [ 
{"producto": "laptop", "tipo": "salida", "cantidad": 2}, 
{"producto": "mouse", "tipo": "entrada", "cantidad": 5}, 
{"producto": "monitor", "tipo": "salida", "cantidad": 6}, 
{"producto": "teclado", "tipo": "entrada", "cantidad": 3} 
] 

#Requisitos
#● No permitas que el stock quede negativo. 
#● Registra los movimientos rechazados en una lista. 
#● Crea productos nuevos solo mediante entradas. 

secuencia_movimientos = []
movimientos_rechazados = []
movimientos_aceptados = []

for movimiento in movimientos:
    producto = movimiento["producto"]
    tipo = movimiento["tipo"]
    cantidad = movimiento["cantidad"]

    if tipo == "entrada":
        if producto in stock:
            stock[producto] += cantidad
        else:
            stock[producto] = cantidad
        movimientos_aceptados.append(movimiento)
    elif tipo == "salida":
        if producto in stock and stock[producto] >= cantidad:
            stock[producto] -= cantidad
            movimientos_aceptados.append(movimiento)
        else:
            movimientos_rechazados.append(movimiento)
            secuencia_movimientos.append(movimiento)
print("movimientos_rechazados:", movimientos_rechazados )

print("movimientos_aceptados:", movimientos_aceptados )

creacion_stock = {}
for movimiento in movimientos_aceptados:
    producto = movimiento["producto"]
    if producto not in creacion_stock:
        creacion_stock[producto] = 0
    creacion_stock[producto] += 1
print("creacion_stock:", creacion_stock )
```

```python
#Ejercicio 10
#10. Matriz dispersa con diccionarios 
#Representa y manipula una matriz dispersa utilizando un diccionario cuyas claves sean tuplas (fila, columna).

matriz = { 
(0, 1): 5, 
(1, 3): 8, 
(2, 0): 4, 
(3, 2): 7 
} 
filas = 4 
columnas = 4 

#Requisitos:
#● Convierte la matriz dispersa en una lista de listas. 
#● Calcula la suma por fila y por columna. 
#● Obtén la matriz transpuesta como otro diccionario. 

matriz_lista = [[0 for _ in range(columnas)] for _ in range(filas)]
for (fila, columna), valor in matriz.items():
    matriz_lista[fila][columna] = valor

suma_filas = [sum(fila) for fila in matriz_lista]
suma_columnas = [sum(matriz_lista[fila][col] for fila in range(filas)) for col in range(columnas)]
print("Matriz como lista de listas:", matriz_lista)
print("Suma por fila:", suma_filas)
print("Suma por columna:", suma_columnas)

matriz_transpuesta = {}
for (fila, columna), valor in matriz.items():
    matriz_transpuesta[(columna, fila)] = valor
print("Matriz transpuesta:", matriz_transpuesta)
```

```python
#Ejercisio 11
#11. Historial de precios 
#Analiza el historial de precios de varios productos. 
historial = { 
"laptop": [3500, 3450, 3600, 3400], 
"mouse": [80, 85, 78, 82], 
"monitor": [900, 920, 910, 880] 
} 

#Requisitos
#● Calcula precio mínimo, máximo, promedio y variación porcentual entre el primer y último valor. 
#● Identifica el producto con mayor reducción porcentual. 

precio_analisis = {}
for producto, precios in historial.items():
    minimo = min(precios)
    maximo = max(precios)
    promedio = sum(precios) / len(precios)
    variacion_porcentual = ((precios[-1] - precios[0]) / precios[0]) * 100
    precio_analisis[producto] = {
        "minimo": minimo,
        "maximo": maximo,
        "promedio": promedio,
        "variacion_porcentual": variacion_porcentual
    }
print("Análisis de precios:", precio_analisis)

producto_mayor_reduccion = max(precio_analisis, key=lambda p: precio_analisis[p]["variacion_porcentual"])
print("Producto con mayor reducción porcentual:", producto_mayor_reduccion)
```

```python
#Ejercisio 12
#12. Detección de registros duplicados 
#Detecta posibles usuarios duplicados utilizando correo o número de documento. 

usuarios = [ 
{"id": 1, "nombre": "Ana", "correo": "ana@mail.com", "documento": "70112233"}, 
{"id": 2, "nombre": "Luis", "correo": "luis@mail.com", "documento": "74445566"}, 
{"id": 3, "nombre": "Ana T.", "correo": "ANA@mail.com", "documento": "70112233"}, 
{"id": 4, "nombre": "Carla", "correo": "carla@mail.com", "documento": "78889900"} 
] 

#Requisitos
#● Normaliza los correos antes de compararlos. 
#● Genera un diccionario que relacione cada tipo de duplicado con los IDs implicados. 

normalizar_correo = lambda correo: correo.strip().lower()
duplicados = {"correo": {}, "documento": {}}
print("Usuarios a analizar:", usuarios)

for usuario in usuarios:
    correo_normalizado = normalizar_correo(usuario["correo"])
    documento = usuario["documento"]

    if correo_normalizado not in duplicados["correo"]:
        duplicados["correo"][correo_normalizado] = []
    duplicados["correo"][correo_normalizado].append(usuario["id"])

    if documento not in duplicados["documento"]:
        duplicados["documento"][documento] = []
    duplicados["documento"][documento].append(usuario["id"])
print("Duplicados detectados:", duplicados)
```

```python
#Ejercicio 14
#14. Total de productos en el carrito 
#Calcula cuántos productos hay en total en un carrito de compras y cuántos productos diferentes contiene. 

carrito = { 
"manzana": 4, 
"pan": 2, 
"leche": 1, 
"huevo": 12 
} 

#requisitos
#● Muestra la cantidad total de unidades sumando todos los valores. 
#● Muestra cuántos productos diferentes hay. 
#● Indica el producto con mayor cantidad. 

total_unidades = sum(carrito.values())
productos_diferentes = len(carrito)
producto_mayor_cantidad = max(carrito, key=carrito.get)

print(f"Cantidad total de unidades: {total_unidades}")
print(f"Número de productos diferentes: {productos_diferentes}")
print(f"Producto con mayor cantidad: {producto_mayor_cantidad} ({carrito[producto_mayor_cantidad]})")
```
